```python
#Latihan 1
import numpy as np

# NumPy Array
harga = np.array([5000, 7000, 3000, 12000, 4500])

print('Rata-rata harga:', harga.mean())
print('Harga tertinggi:', harga.max())
print('Harga setelah diskon 10%:', harga * 0.9)

# Perbandingan dengan list Python biasa

harga_list = [5000, 7000, 3000]

print('List dikali 2:', harga_list * 2)

try:
    print('List dikali 0.9:', harga_list * 0.9)
except TypeError:
    print('List tidak bisa dikalikan dengan angka desimal')
```

Tugas Analisis 1

Pertanyaan: Bandingkan harga * 0.9 dengan jika harga berupa list Python biasa. Apa yang terjadi bila mencoba [5000,7000,3000] * 0.9? Jelaskan mengapa berbeda.

Jawaban:
Kalau menggunakan NumPy array, harga * 0.9 bisa langsung menghitung diskon untuk semua harga. Sedangkan kalau menggunakan list Python biasa, [5000,7000,3000] * 0.9 akan menghasilkan error karena list tidak bisa dikalikan dengan angka desimal. NumPy memang dibuat untuk mempermudah operasi angka pada banyak data sekaligus.

```python
#Latihan 3
import pandas as pd

# Data Loading
df = pd.read_csv('data_kantin.csv')

# Data Inspection
print(df.head())# 5 baris pertama
print(df.info())# tipe data & jumlah non-null tiap kolom
print(df.describe())# statistik ringkas kolom numerik
print(df.shape)# jumlah (baris, kolom)

# Mengecek jumlah data yang terisi di setiap kolom

print('Jumlah data:')
print(len(df))

print('\nJumlah data yang terisi:')
print(df.notnull().sum())

print('\nJumlah data kosong:')
print(df.isnull().sum())
```

Tugas Analisis 3

Pertanyaan: Dari hasil df.info(), kolom mana yang jumlah non-null-nya lebih sedikit dari jumlah baris total? Apa artinya?

Jawaban:
Kolom yang jumlah non-null-nya lebih sedikit dari jumlah baris berarti kolom tersebut memiliki data kosong atau missing value. Jadi tidak semua baris pada kolom tersebut mempunyai data.

```python
#Latihan 4
# Mengecek data kosong
print(df.isnull().sum())

# Mengisi data terjual yang kosong dengan 0
df['terjual'] = df['terjual'].fillna(0)

# Menghapus baris yang menu-nya kosong
df = df.dropna(subset=['menu'])

print(df)
```

Jawaban:
Kolom terjual diisi dengan 0 karena data kosong bisa dianggap tidak ada penjualan. Sedangkan baris dengan menu kosong dihapus karena nama menu merupakan informasi penting untuk mengetahui data tersebut adalah penjualan apa. Kalau menu-nya tidak ada, data tersebut sulit digunakan dalam analisis.

```python
#Latihan 5
# Mengecek jumlah data duplikat
print(df.duplicated().sum())

# Menghapus data duplikat
df = df.drop_duplicates()

# memastikan tipe data harga adalah integer
df['harga'] = df['harga'].astype(int)

print(df.dtypes)
```

Dugaan Masalah Kualitas Data:

- Missing value pada kolom guru_pengampu
- Duplikat data
- Tipe data nilai tidak konsisten karena terdapat data seperti "67 poin"

Rencana Teknik Pembersihan:
- Mengubah nilai menjadi numerik dan menghapus teks "poin"
- Nilai kosong diisi menggunakan median
- Nilai di luar rentang 0-100 dianggap tidak valid lalu diperbaiki
- Guru yang kosong diisi dengan "Belum diisi"
- Data duplikat dihapus
- Format tanggal diseragamkan menjadi datetime

Rencana Manipulasi Data:
Filter: nilai >= 75
Sort: nilai dari terbesar ke terkecil
Kolom turunan: status_nilai
Groupby/agregasi: rata-rata nilai berdasarkan mata pelajaran



```python
import pandas as pd
import numpy as np

# Data Loading
df = pd.read_csv('dataset_nilai_akademik_siswa.csv')

print(df.head())
```

```python
# Data Inspection

print("Info Dataset:")
print(df.info())

print("\nStatistik Dataset:")
print(df.describe(include='all'))

print("\nUkuran Dataset:")
print(df.shape)
```

```python
# Cek Missing Value
print("Missing Value:")
print(df.isnull().sum())

# Cek Duplikat
print("\nJumlah Duplikat:")
print(df.duplicated().sum())

# Cek Tipe Data
print("\nTipe Data:")
print(df.dtypes)
```

```python
# Cleaning nilai

df['nilai'] = df['nilai'].astype(str).str.replace(' poin', '', regex=False)
df['nilai'] = pd.to_numeric(df['nilai'], errors='coerce')

print(df['nilai'].head())
```

```python
# Cek nilai di luar rentang 0-100

print(df[df['nilai'] > 100][['nama', 'mata_pelajaran', 'nilai']])
```

```python
# Nilai harus berada di antara 0 sampai 100

df.loc[(df['nilai'] < 0) | (df['nilai'] > 100), 'nilai'] = np.nan

# Mengisi nilai kosong dengan median
df['nilai'] = df['nilai'].fillna(df['nilai'].median())
```

```python
# Mengisi guru yang kosong

df['guru_pengampu'] = df['guru_pengampu'].fillna('Belum diisi')
```

```python
# Menyamakan penulisan jenis ujian

df['jenis_ujian'] = df['jenis_ujian'].str.upper()
```

```python
# Menyamakan format tanggal

bulan = {
    'Januari': '01',
    'Februari': '02',
    'Maret': '03',
    'April': '04',
    'Mei': '05',
    'Juni': '06',
    'Juli': '07',
    'Agustus': '08',
    'September': '09',
    'Oktober': '10',
    'November': '11',
    'Desember': '12'
}

tanggal = df['tanggal_ujian']

for nama_bulan, nomor_bulan in bulan.items():
    tanggal = tanggal.str.replace(nama_bulan, nomor_bulan, regex=False)

df['tanggal_ujian'] = pd.to_datetime(
    tanggal,
    format='mixed',
    dayfirst=True,
    errors='coerce'
)
```

```python
# Hapus data duplikat

df = df.drop_duplicates()

print("Ukuran data setelah cleaning:", df.shape)
```

```python
# Cek ulang dataset

print(df.info())

print("\nMissing Value:")
print(df.isnull().sum())

print("\nJumlah Duplikat:")
print(df.duplicated().sum())

print("\nData:")
print(df.head())
```

```python
# Filtering

nilai_tinggi = df[df['nilai'] >= 75]

print(nilai_tinggi)
```

```python
# Sorting

nilai_urut = df.sort_values(by='nilai', ascending=False)

print(nilai_urut[['nama', 'mata_pelajaran', 'nilai']])
```

```python
# Kolom turunan

df['status_nilai'] = np.where(
    df['nilai'] >= 75,
    'Baik',
    'Perlu Peningkatan'
)

print(df[['nama', 'nilai', 'status_nilai']].head())
```

```python
# Groupby

rata_nilai_mapel = df.groupby('mata_pelajaran')['nilai'].mean().sort_values(
    ascending=False
)

print(rata_nilai_mapel)
```

```python
# Rata-rata nilai berdasarkan kelas

rata_nilai_kelas = df.groupby('kelas')['nilai'].mean().sort_values(
    ascending=False
)

print(rata_nilai_kelas)
```

```python
# Rata-rata nilai berdasarkan jenis ujian

rata_nilai_ujian = df.groupby('jenis_ujian')['nilai'].mean().sort_values(
    ascending=False
)

print(rata_nilai_ujian)
```
